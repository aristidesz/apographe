```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Read and concatenate all CSVs in the data folder
csv_files = list(Path("../data").glob("fuel_prices_*.csv"))
df = pd.concat([pd.read_csv(f) for f in csv_files], ignore_index=True)

# Ensure datetime is parsed as datetime
df["datetime"] = pd.to_datetime(df["datetime"])

# Define sites to plot (brand, address)
sites = [
    ("PETROLINA", "Λεωφ. Ελλάδος 65 Τ.Κ 7101\n τηλ: 24323063"),
    ("FILL N GO", "Λεωφ. Ελλάδος 76 Τ.Κ 7101\n τηλ: 24533460"),
    ("SHELL", "Λεωφ. Ελλάδος 76 Τ.Κ 7101\n τηλ: 24533460"),
    ("EKO", "Κυριάκου Μάτση 100 Τ.Κ 7102\n τηλ: 24532888"),
]

fuel_type = "Αμόλυβδη 95"

# Plot price history for both sites on the same figure
plt.figure(figsize=(10, 5))
for site_brand, site_address in sites:
    site_df = df[
        (df["brand"] == site_brand)
        & (df["address"] == site_address)
        & (df["fuel_type"] == fuel_type)
    ]
    site_df = site_df.sort_values("datetime")
    plt.plot(
        site_df["datetime"], site_df["price"], label=f"{site_brand} - {site_address}"
    )

plt.title(f"Price history for selected sites ({fuel_type})")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
